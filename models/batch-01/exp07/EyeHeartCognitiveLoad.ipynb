### Konfigurasi Eksperimen

Menetapkan dataset, seed, direktori hasil, serta mode evaluasi. `full` mengevaluasi seluruh peserta; `smoke` membatasi percobaan untuk pemeriksaan teknis. `RESUME_DIR` menentukan apakah runner membuat run baru atau melanjutkan checkpoint yang cocok.

```python
import json
import pandas as pd
from IPython.display import display, Image

sys.path.insert(0, str(MODELS_DIR))
from shared.experiments.runner import Config, run
from shared.experiments.design import registry
from shared.data import load_dataset, describe_data, make_folds

EXPERIMENT = 'exp07'
MODE = 'full'
SEED = 42
RESUME_DIR = None
DATASET_PATH = ROOT / 'Dataset' / 'good_acceptable.csv'
OUTPUT_ROOT = MODELS_DIR / 'outputs'
config = Config(DATASET_PATH, OUTPUT_ROOT, EXPERIMENT, MODE, SEED)
```

### Dataset dan Kandidat SMOTE

Memuat dataset dan memeriksa distribusi kelas per peserta. Fitur `clean` menghapus `blink_count`, `iris_size_delta_std`, dan `eye_state_other_ratio` karena redundansi atau nilai konstan yang sudah ditemukan. Kedua model memakai fitur clean dan StandardScaler, sama dengan kontrol pada exp05 dan exp06.

Uji target Low sebesar 50%, 75%, atau 100% dari jumlah kelas terbesar training, dengan k_neighbors 3 dan 5. Ada enam kombinasi per model pada mode full. Target memakai pembulatan ke atas dan tidak menurunkan Low yang sudah mencukupi. Medium dan High tetap; validation/test tidak di-resample.

```python
data = load_dataset(DATASET_PATH)
print(describe_data(data))
display(pd.crosstab(data.participant_id, data.difficulty_level))
folds = make_folds(data)
specs = registry(EXPERIMENT, MODE, SEED)
display(pd.DataFrame([{'model': s['id'], 'features': s['variant'], 'scaler': s['scaler'], 'candidates': len(s['candidates']), 'inner_search': s['tune']} for s in specs]))
print('Outer folds:', len(folds))
```

### Scaling, SMOTE, dan Pemilihan Kandidat

Setiap peserta menjadi data uji sekali, sementara data peserta lain digunakan untuk training. Identitas peserta hanya dipakai untuk pembagian data, bukan fitur prediksi.

Pada setiap inner training, hapus fitur kosong, fit imputasi median, hapus fitur konstan, lalu fit StandardScaler. SMOTE mencari tetangga dan membuat contoh Low pada fitur yang sudah diskalakan. Pemeriksaan nilai sintetis memastikan nilai finite dan tetap dalam batas per fitur training.

Pilih kombinasi rasio dan k dengan macro-F1 gabungan tiga inner validation. Kandidat yang kekurangan contoh Low untuk k tertentu tidak mendapat skor parsial atau fallback ke random oversampling. Setelah pemilihan, fit ulang seluruh pipeline pada outer training dan prediksi peserta uji. Kontrol tetap tanpa resampling.

```python
run_dir = run(config, resume_dir=RESUME_DIR)
```

### Evaluasi SMOTE

Macro-F1 dari gabungan prediksi outer test menjadi metrik utama. Accuracy dan balanced accuracy melengkapi precision, recall, dan F1 tiap kelas. Hasil per peserta memperlihatkan apakah performa merata atau hanya baik pada peserta tertentu. Nilai kosong pada metrik kelas berarti peserta uji tidak memiliki contoh kelas tersebut.

Bandingkan varian `balanced` terhadap `control` pada model yang sama. Untuk membandingkan SMOTE dengan exp06, gunakan run full dengan data, manifest, environment, dan model yang cocok. Periksa F1/precision/recall Low serta perubahan kelas lain.

`search_results.csv` mencatat kandidat yang tidak layak dan `selected_candidates.json` mencatat pilihan per outer fold. Contoh sintetis yang lolos pemeriksaan rentang tidak otomatis merupakan rekaman fisiologis nyata atau menggantikan keragaman peserta baru.

```python
summary = json.loads((run_dir / 'summary.json').read_text())
print(summary)
if summary['status'] != 'complete':
    raise RuntimeError('Incomplete evaluation')
if MODE == 'smoke':
    print('Smoke test only. Change MODE to full and run again for research evaluation.')
display(pd.read_csv(run_dir / 'metrics.csv'))
classes = pd.read_csv(run_dir / 'class_metrics.csv')
display(classes[classes.scope == 'pooled'])
display(pd.read_csv(run_dir / 'participant_metrics.csv'))
display(pd.read_csv(run_dir / 'search_results.csv'))
print('Saved results:', run_dir)
```
